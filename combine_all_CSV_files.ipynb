```python
import os
import pandas as pd

def combine_csv_from_folder(folder_path):
    all_data = []

    # Loop through all CSV files in the folder
    for file in os.listdir(folder_path):
        if file.endswith(".csv") and file != "combined_CSV_files.csv":  # avoid re-reading the output file
            file_path = os.path.join(folder_path, file)
            
            # Read CSV (only two rows: header + one row of data)
            df = pd.read_csv(file_path)
            
            if df.shape[0] != 1:
                print(f"Warning: {file} has {df.shape[0]} data rows, only taking the first row.")
                df = df.iloc[[0]]  # keep only the first row of data
            
            # Add a column for the file name (insert at first position)
            df.insert(0, "source_file", file)
            
            all_data.append(df)

    # Concatenate all into one DataFrame
    if all_data:
        combined_df = pd.concat(all_data, ignore_index=True)
        
        # Save to CSV in the same folder
        output_path = os.path.join(folder_path, "combined_CSV_files.csv")
        combined_df.to_csv(output_path, index=False)
        print(f"Combined CSV saved to: {output_path}")
        
        return combined_df
    else:
        print("No CSV files found in the folder.")
        return pd.DataFrame()

if __name__ == "__main__":
    folder = input("Enter the folder path containing the CSV files: ").strip()
    result_df = combine_csv_from_folder(folder)
    
    if not result_df.empty:
        print("Combined DataFrame:")
        print(result_df)
```
